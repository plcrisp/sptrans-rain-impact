## Extração das Rotas SPTrans e Estações Pluviométricas do CEMADEN

```python
import sys
import os
import zipfile
from pathlib import Path
from IPython.display import display

sys.path.append(os.path.abspath('..'))

from src.utils.gtfs_parser import build_gtfs_dataset
from src.core.config import Config

# Extrai automaticamente o gtfs.zip caso os arquivos ainda não tenham sido extraídos
gtfs_dir = Path(Config.GTFS_DATA_DIR)
gtfs_zip = gtfs_dir / 'gtfs.zip'

if gtfs_zip.exists() and not (gtfs_dir / 'shapes.txt').exists() and not (gtfs_dir / 'shapes.csv').exists():
    print(f'Extraindo {gtfs_zip.name} em {gtfs_dir}...')
    with zipfile.ZipFile(gtfs_zip, 'r') as z:
        z.extractall(gtfs_dir)
    print('Extração concluída com sucesso!')

df = build_gtfs_dataset(Config.GTFS_DATA_DIR, Config.GTFS_DATA_DIR)

display(df.head())
```
